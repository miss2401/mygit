```python
import numpy as np
```

```python
#Tự đi mà đọc đề
x1 = np.array([3, 4, 2, 6, 8, 2, 5])
x2 = np.array([5, 5.5, 4, 7, 10, 5, 7.5])
n = len(x1)
print(x1, "\n", x2)
```

```python
x_1 = np.sum(x1)/n
x_2 = np.sum(x2)/n
print("Mean x1 = ", x_1)
print("Mean x2 = ", x_2)
```

```python
s11 = np.sum((x1 - x_1)**2)/(n - 1)
s22 = np.sum((x2 - x_2)**2)/(n - 1)
s12 = np.sum((x1 - x_1)*(x2 - x_2)) /(n - 1)
S = np.array([[s11, s12], [s12, s22]])
print("Covariance matrix: ", S)
```
